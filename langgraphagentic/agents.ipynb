# IngredientSight AI - LangGraph Agentic Pipeline

This notebook demonstrates the full LangGraph pipeline with 5 agent nodes:

```
START → OCR → Ingredient → Research → Safety → Report → END
```

## Agent 1: OCR Agent (Gemini Vision)

````python
from langgraphagentic.ocr_agent import ocr_node
````

## Agent 2: Ingredient Agent (LangChain + Gemini)

````python
from langgraphagentic.ingredient_agent import ingredient_node
````

## Agent 3: Research Agent (DuckDuckGo + Gemini)

````python
from langgraphagentic.research_agent import research_node
````

## Agent 4: Safety Agent (LangChain + Gemini)

````python
from langgraphagentic.safety_agent import safety_node
````

## Agent 5: Report Agent (Markdown + JSON Generator)

````python
from langgraphagentic.report_agent import report_node
````

---
## Build the LangGraph StateGraph

All 5 agents are registered as **nodes** and connected via **edges** in a `StateGraph`.

````python
from langgraphagentic.graph import build_graph, PipelineState

# Build and compile the graph
app = build_graph()
print("LangGraph compiled successfully!")
print("Topology: START -> ocr -> ingredient -> research -> safety -> report -> END")
````

### Visualize the Graph

````python
# Print the graph structure as ASCII
try:
    print(app.get_graph().draw_ascii())
except Exception:
    print("ASCII visualization not available. Graph nodes:", list(app.get_graph().nodes.keys()))
````

---
## Run the Pipeline

Provide an image path and invoke the entire graph with a single call.

````python
# Invoke the LangGraph pipeline
result = app.invoke({"image_path": "ingredients_en.5.full.jpg"})

# Display results
print("OCR Text:", result.get("ocr_text"))
print("Ingredients:", result.get("ingredients"))
print("Safety Score:", result.get("safety_analysis", {}).get("overall_score"))
print("Markdown Report:", result.get("report_md_path"))
print("JSON Report:", result.get("report_json_path"))
````
